## Different Approaches to AI

#### 1. Rule based
In rule based systems, we explicitly define conditions and actions.

````python
def check_weather(temp):
    if temp > 35:
        return "It is very hot today"

    elif temp > 25:
        return "The weather is warm"

    else:
        "The weather is cool"
````

#### 2. Machine Learning based AI
Instead of manually defining every rule, we provide data to and algorithm so that it can identify patterns and learn from example.

#### 3. Deep learning based AI

Deep learning is a subset of ML that uses artificial neural networks with multiple layers to learn complex patterns.

It is particularly useful for:

- image recognition
- speech recognition
- Natural language processing
- Autonomous Vehicles.
- Generative AI


#### 4. Generative AI

Generative AI that learn patterns from training data and generate new content.

Example:

- Generatinge Article
- Writing Python code
- Creating images
- Generating music
- Summarizing documents
- Answering Questions

# Generative AI

GenAI uses trained model on large datasets to generate new content based on patterns learned during training.


#### What is token?

A token is unit of text processed by a language model

A token can represent"
- entire word.
- part of word.
- punctuation mark.
- a number.
- a special symbol.

"Generative AI is amazing"

A tokenizer might split into tokens approximately like.
["Generative","AI","is","amazing"]

#### What is large language model(LLM)s

A Large language models is a type of AI model trained on large amount of text data to understand and generate human like language.

Examples:

- GPT
- gemini
- Claude
- Llama
etc.

At high level, an LLM works through stages:

1. Input: Receive a user prompt
2. Tokenization: Converts the text into tokens.
3. Embedding: Converts tokens into numrical representation.
4. Processing: Neural network layers process contextual relationships.
5. Prediction: Predicts the next token.
6. Generation: Repeats prediction to generate response.


#### What is a prompt ?

A prompt is the instruction or input that we provide to a generative AI model.



````python
from groq import Groq
from dotenv import load_dotenv
import os


load_dotenv()

client = Groq(
    api_key = os.environ.get("GROQ_API_KEY")
)

response = client.chat.completions.create(
    messages=[
        {
            "role":"user",
            "content":"Explain the AI vs ML vs DL vs GenerativeAI"
        },
    ],
    model="openai/gpt-oss-20b"
)
````

## AI, ML, DL, and Generative AI – A “Road‑Map” Guide  

| Term | What it means | Typical techniques | Key goal | Examples |
|------|---------------|-------------------|----------|----------|
| **Artificial Intelligence (AI)** | A broad umbrella that covers *any* system that can perform tasks that would normally require human intelligence. | Rule‑based systems, expert systems, symbolic reasoning, machine learning, deep learning, evolutionary algorithms, etc. | Mimic or surpass human decision‑making, perception, or creativity. | Chess programs, autonomous cars, virtual assistants, game‑AI. |
| **Machine Learning (ML)** | A subset of AI where the system *learns* from data instead of being hand‑coded for every case. | Regression, classification, clustering, reinforcement learning, decision trees, support vector machines, neural networks, etc. | Extract patterns, make predictions or decisions without explicit programming. | Spam filters, recommendation engines, fraud detection. |
| **Deep Learning (DL)** | A subset of ML that uses *deep neural networks* (many layers) to learn hierarchical representations from data. | Convolutional neural networks (CNNs), recurrent/transformer networks, generative adversarial networks (GANs), diffusion models, etc. | Automate feature extraction and achieve state‑of‑the‑art performance on perception tasks. | Image recognition (ImageNet), speech recognition, language translation. |
| **Generative AI** | AI that *creates* new data (text, image, audio, video, code) rather than just classifying or predicting. Usually built on DL generative models. | Transformers (GPT, T5), diffusion models (Stable Diffusion, Imagen), GANs, autoregressive models. | Produce novel, high‑quality content that is useful or entertaining. | ChatGPT, DALL‑E, Midjourney, music generators, synthetic training data. |

## 1. AI vs ML: Scope & Philosophy

| | **Artificial Intelligence** | **Machine Learning** |
|---|---|---|
| **Definition** | Any method that lets machines act intelligently. | A specific approach that lets machines *learn* from data. |
| **Paradigm** | Can be *symbolic* (logic rules, knowledge graphs) or *subsymbolic* (learning from data). | Purely sub‑symbolic, driven by statistical inference. |
| **Goal** | Emulate human cognitive abilities (reasoning, planning, learning, perception, language). | Learn predictive or decision‑making patterns from data. |
| **Examples** | Expert system “MYCIN”, rule‑based chess AI, classical planning algorithms. | Linear regression, decision trees, k‑means clustering, support‑vector machines. |
| **Typical Question** | “Can a machine *think*?” | “Can a machine *learn* from examples?” |

> **TL;DR** – *AI* is the umbrella, *ML* is a powerful sub‑field within that umbrella.

---

## 2. ML vs DL: Depth & Data

| | **Machine Learning** | **Deep Learning** |
|---|---|---|
| **Model Complexity** | Shallow, often a single or a few layers. | Very deep (tens to hundreds of layers). |
| **Feature Engineering** | Human‑designed features (e.g., hand‑crafted image descriptors). | Automatic feature learning; network learns representation. |
| **Data Needs** | Can work well with modest data (few‑shot). | Requires large amounts of labeled data (thousands–millions). |
| **Computation** | Usually lighter (CPU, modest GPU). | Heavy GPU/TPU compute, long training times. |
| **Explainability** | Often easier (coefficients, trees). | Harder (“black box”). |
| **Typical Tasks** | Small‑scale classification, regression, clustering. | Image classification, object detection, natural language generation, etc. |
| **Key Models** | Random forests, SVMs, k‑NN, logistic regression. | CNNs, LSTMs, transformers, autoencoders, GANs, diffusion models. |

> **TL;DR** – DL is *ML with depth*. It’s the engine behind most state‑of‑the‑art perception and generative systems.

---

## 3. Generative AI – Turning Data Into Creativity

### 3.1 What Makes a Model “Generative”?

| | **Discriminative** | **Generative** |
|---|---|---|
| **Goal** | Predict label/response given input. | Generate realistic samples from a learned distribution. |
| **Typical Tasks** | Classification, regression, ranking. | Image synthesis, text completion, music composition, data augmentation. |
| **Model Types** | Logistic regression, SVM, CNN classifiers. | GANs, VAEs, autoregressive transformers, diffusion models. |

### 3.2 How Do Generative Models Work?

| Model | Core Idea | Key Architecture | Typical Use |
|-------|-----------|------------------|-------------|
| **GAN (Generative Adversarial Network)** | Two networks: generator vs discriminator; adversarial training. | Deep CNN generator + CNN discriminator. | High‑fidelity image generation, style transfer. |
| **VAE (Variational Autoencoder)** | Probabilistic latent space + reconstruction loss. | Encoder → latent → decoder. | Controlled image editing, latent manipulation. |
| **Transformers (autoregressive)** | Predict next token conditioned on previous ones (text, music, etc.). | Self‑attention layers + positional encodings. | Language models (GPT‑3/4), code generation. |
| **Diffusion Models** | Add noise to data then learn to reverse diffusion. | U‑Net with timestep conditioning. | Photo‑realistic image generation (Stable Diffusion, Imagen). |

### 3.3 Why the “AI” vs “ML” Distinction?

- **AI**: *What* a system can do (e.g., produce a novel painting).
- **ML**: *How* it learns the rules to do it (e.g., learning from millions of images).
- **DL**: The *tool* that makes the learning feasible and efficient for complex, high‑dimensional data.
- **Generative AI**: The *application domain* (creative content generation) that usually relies on DL.

## 4. Quick “Decision Tree” to Know Where You’re At

```
Does your system use data to improve?
     |
     +-- No  --> Rule‑based AI, symbolic reasoning
     |
     +-- Yes --> Machine Learning
                 |
                 +-- Deep Neural Networks? --> Deep Learning
                 |        |
                 |        +-- Is the goal to *create* new data? --> Generative AI
                 |
                 +-- Is it a simple model or a shallow network? --> Traditional ML
```

---

## 5. Concrete Example: From AI → ML → DL → Generative AI

1. **AI Goal**: Build a virtual assistant that can hold conversations and create images on demand.
2. **ML Solution**: Train a language model (supervised learning) on dialog datasets to respond to queries.
3. **DL Enhancement**: Use a transformer (deep network) to capture long‑range dependencies and generate coherent text.
4. **Generative Extension**: Add a diffusion model that can produce images from textual prompts—now the assistant *creates* pictures in addition to text.
