## Different Approaches to AI

#### 1. Rule based
In rule based systems, we explicitly define conditions and actions.

```python
def check_weather(temp):
    if temp > 35:
        return "It is very hot today"

    elif temp > 25:
        return "The weather is warm"

    else:
        "The weather is cool"
```

#### 2. Machine Learning based AI
Instead of manually defining every rule, we provide data to and algorithm so that it can identify patterns and learn from example.

#### 3. Deep learning based AI

Deep learning is a subset of ML that uses artificial neural networks with multiple layers to learn complex patterns.

It is particularly useful for:

- image recognition
- speech recognition
- Natural language processing
- Autonomous Vehicles.
- Generative AI


#### 4. Generative AI

Generative AI that learn patterns from training data and generate new content.

Example:

- Generatinge Article
- Writing Python code

- Generating music
- Summarizing documents
- Answering Questions

# Generative AI

GenAI uses trained model on large datasets to generate new content based on patterns learned during training.


#### What is token?

A token is unit of text processed by a language model

A token can represent"
- entire word.
- part of word.
- punctuation mark.
- a number.
- a special symbol.

"Generative AI is amazing"

A tokenizer might split into tokens approximately like.
["Generative","AI","is","amazing"]

#### What is large language model(LLM)s

A Large language models is a type of AI model trained on large amount of text data to understand and generate human like language.

Examples:

- GPT
- gemini
- Claude
- Llama
etc.

At high level, an LLM works through stages:

1. Input: Receive a user prompt
2. Tokenization: Converts the text into tokens.
3. Embedding: Converts tokens into numrical representation.
4. Processing: Neural network layers process contextual relationships.
5. Prediction: Predicts the next token.
6. Generation: Repeats prediction to generate response.


#### What is a prompt ?

A prompt is the instruction or input that we provide to a generative AI model.



```python
from groq import Groq
from dotenv import load_dotenv
import os


load_dotenv()

client = Groq(
    api_key = os.environ.get("GROQ_API_KEY")
)

response = client.chat.completions.create(
    messages=[
        {
            "role":"user",
            "content":"Explain the AI vs ML vs DL vs GenerativeAI"
        },
    ],
    model="openai/gpt-oss-20b"
)
```

### AI → ML → DL → Generative AI  
A quick‑look hierarchy:

| Level | Scope | Typical techniques | What it *does* | Example |
|-------|-------|--------------------|----------------|---------|
| **Artificial Intelligence (AI)** | Broadest concept – systems that can **perform tasks that normally require human intelligence** (reasoning, perception, planning, etc.) | Rule‑based engines, expert systems, symbolic reasoning, ML, DL, etc. | *Can I play chess?* | Chess‑playing engine (e.g., Stockfish) |
| **Machine Learning (ML)** | AI that **learns from data** instead of being hard‑coded | Decision trees, SVMs, k‑NN, random forests, neural nets, etc. | *What’s the sentiment in a tweet?* | Naïve Bayes text classifier |
| **Deep Learning (DL)** | ML that uses **deep (many‑layer) neural networks** | Feed‑forward nets, CNNs, RNNs, transformers, GANs, VAEs, etc. | *Recognize a cat in a photo.* | ResNet‑50 image classifier |
| **Generative AI** | DL (often) that **creates new data** (text, image, audio, video) that looks realistic | Diffusion models, GANs, VAEs, transformer‑based language models, etc. | *Write a poem that sounds like Shakespeare.* | GPT‑4, Stable Diffusion, Midjourney |

## 1. Artificial Intelligence (AI)

- **Goal**: Build systems that can act intelligently or autonomously.
- **Approaches**:  
  - *Symbolic/Rule‑based*: If‑then logic, expert systems.  
  - *Statistical*: Machine learning.  
  - *Hybrid*: Combine symbolic + learning (e.g., neuro‑symbolic systems).
- **Typical tasks**: Vision, NLP, robotics, planning, recommendation, games, etc.
- **Not always data‑driven**: You can hand‑craft knowledge bases.

### Quick Example  
A virtual assistant (Siri, Alexa) that understands spoken commands, fetches data, and performs actions.

---

## 2. Machine Learning (ML)

- **Subset of AI** that relies on data‑driven pattern recognition.
- **Core idea**: Learn a function *f* from input *x* to output *y* using a dataset *D = {(x_i, y_i)}*.
- **Learning paradigms**:
  - **Supervised**: Known labels (classification, regression).  
  - **Unsupervised**: No labels (clustering, dimensionality reduction).  
  - **Reinforcement**: Learn by interacting with an environment (reward signals).  
  - **Self‑supervised / Contrastive**: Generate labels from raw data (used a lot in modern NLP/vision).
- **Examples**:
  - *Predict house price* (regression).  
  - *Detect spam* (classification).  
  - *Group customers by purchase habits* (clustering).

### Key Point  
ML is the **“tool”**—the algorithmic mechanism—inside many AI systems. It can be shallow (e.g., linear regression) or deep.

---

## 3. Deep Learning (DL)

- **Subset of ML** that uses *deep* neural networks: many hidden layers.
- **Why “deep”?**: More layers = greater capacity to model complex, hierarchical patterns.
- **Common architectures**:
  - **CNN** (Convolutional Neural Networks) – images, video.  
  - **RNN / LSTM / GRU** – sequences (time‑series, speech).  
  - **Transformer** – self‑attention, excels in long‑range dependencies (NLP, vision).  
  - **GAN (Generative Adversarial Network)** – generative tasks.  
  - **VAE (Variational Auto‑Encoder)** – generative tasks.  
  - **Diffusion Models** – recent image/audio generation.
- **Training**: Stochastic gradient descent + back‑propagation.  
- **Data needs**: Usually *large* datasets; can use transfer learning or pre‑training.
- **Typical tasks**: Object detection, speech recognition, translation, style transfer, etc.

### Quick Example  
A CNN trained on millions of labeled photos learns to detect cats and dogs.

---

## 4. Generative AI

- **Definition**: AI that *generates* new, synthetic data rather than just classifying or predicting.
- **Key property**: The model learns the *distribution* of data, p(x), so it can sample from it.
- **Typical generative models**:
  - **GANs**: Two networks (generator + discriminator) train in competition.  
  - **VAEs**: Encode data into a latent space, then decode to reconstruct.  
  - **Diffusion Models**: Gradually denoise random noise into realistic data.  
  - **Transformer‑based language models (e.g., GPT‑4)**: Autoregressive models that generate coherent text.  
  - **Diffusion‑based image models (e.g., Stable Diffusion)**.
- **Distinction from discriminative models**:  
  - *Discriminative*: Learn p(y|x) (classify).  
  - *Generative*: Learn p(x) or p(x|y) (create).
- **Typical use‑cases**:  
  - Text generation (stories, code, chat).  
  - Image synthesis (art, design).  
  - Audio/music creation.  
  - Data augmentation for training other models.  
  - Simulation and virtual environments.
- **Why it matters**: Enables *creative* and *productive* applications—drafting, brainstorming, rapid prototyping.

### Quick Example  
You give GPT‑4 a prompt, and it writes a short story in Shakespeare’s style. Or you give a text prompt to Stable Diffusion, and it produces a high‑resolution illustration.

---

## Putting It All Together

1. **AI** is the grand umbrella: any system that acts intelligently.  
2. **ML** is a *data‑driven* way to create AI.  
3. **DL** is ML that uses deep neural nets, giving state‑of‑the‑art performance on many tasks.  
4. **Generative AI** is a special DL area focused on creating new content by learning the underlying distribution of data.
