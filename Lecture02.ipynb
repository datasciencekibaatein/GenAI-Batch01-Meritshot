```python
import tiktoken

text = "We are learning GenAI"

encoding = tiktoken.get_encoding("cl100k_base")

tokens = encoding.encode(text)

print("Token IDs: ",tokens)
print("Total Tokens: ",len(tokens))

decoded_text = encoding.decode(tokens)

print("Decoded Text: ",decoded_text)
```

## Context Window

A context window is the maximum amount of tokenized information a model can consider within a particular request.

It Can include:

- System Instructions
- User Prompt
- Previous conversation message 
- Retrieved documents 
- Generated response tokens.

### Context window vs Memory

- Context Window: Information supplied to the model in its current context.
- Memory: Information stored or managed separately and potentially supplied in future interaction.
